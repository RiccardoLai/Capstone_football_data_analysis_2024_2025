```python
import pandas as pd
import os

# PERCORSO FILE
path = r"C:\Users\bello\Desktop\CAPSTONE\Capstone_calcio_2425\players_raw"

# CARICAMENTO FILE COMPLETO 
players_all = pd.read_csv(os.path.join(path, "players_data_light-2024_2025.csv"))

# AGGIUNGO COLONNA SEASON 
players_all["season"] = "2024-2025"

# CONTROLLO DIMENSIONI
print("Dimensioni dataset:", players_all.shape)

# ANTEPRIMA
print(players_all.head())

# SALVATAGGIO FILE PULITO
output = r"C:\Users\bello\Desktop\CAPSTONE\Capstone_calcio_2425\players_clean\players_all_2024_2025.csv"

players_all.to_csv(output, index=False)

print("FILE CREATO in:")
print(output)
```

```python
# ============================================
# FILTRO SOLO I 3 CAMPIONATI 
# ============================================

players_capstone = players_all[
    players_all["Comp"].isin([
        "it Serie A",
        "es La Liga",
        "eng Premier League"
    ])
]

# ============================================
# CONTROLLO
# ============================================

print("Dimensioni:", players_capstone.shape)

print("\nCampionati presenti:")
print(players_capstone["Comp"].value_counts())

# ============================================
# SALVATAGGIO FILE NELLA CARTELLA PLAYERS CLEAN
# ============================================

# ====== OUTPUT: CAMBIA QUI DOVE SALVI ======
path_out = r"C:\Users\bello\Desktop\CAPSTONE\EXPORT_SICURO"
os.makedirs(path_out, exist_ok=True)

output_file = os.path.join(path_out, "players_capstone_3_leagues_2024_2025.csv")
players_capstone.to_csv(output_file, index=False, encoding="utf-8")

print("\n✅ FILE SALVATO QUI:")
print(output_file)
```
