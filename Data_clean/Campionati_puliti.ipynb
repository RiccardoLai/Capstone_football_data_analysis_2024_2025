```python
# IMPORT LIBRERIA E PERCORSO FILE GREZZI

import pandas as pd

PATH_RAW = r"C:\Users\bello\Desktop\CAPSTONE\Capstone_calcio_2425\data_raw (dati_originali)"
```

```python
# CARICAMENTO FILE GREZZI

serie_a_raw = pd.read_csv(PATH_RAW + r"\matches_serie_a.csv")
premier_raw = pd.read_csv(PATH_RAW + r"\matches_premier_league.csv")
liga_raw = pd.read_csv(PATH_RAW + r"\matches_liga.csv")
```

```python
# ANTEPRIMA DATI GREZZI

print("SERIE A GREZZO")
display(serie_a_raw.head())

print("PREMIER GREZZO")
display(premier_raw.head())

print("LIGA GREZZO")
display(liga_raw.head())
```

```python
# SELEZIONE STESSE 10 COLONNE

COLONNE = ["Date","HomeTeam","AwayTeam","FTHG","FTAG","FTR","HS","AS","HST","AST"]

serie_a_clean = serie_a_raw[COLONNE]
premier_clean = premier_raw[COLONNE]
liga_clean = liga_raw[COLONNE]
```

```python
# ANTEPRIMA DATI FORMATTATI

print("SERIE A CLEAN")
display(serie_a_clean.head())

print("PREMIER CLEAN")
display(premier_clean.head())

print("LIGA CLEAN")
display(liga_clean.head())
```

```python
# PERCORSO CARTELLA DATI PULITI

PATH_CLEAN = r"C:\Users\bello\Desktop\CAPSTONE\Capstone_calcio_2425\data_clean (dati_puliti)"
```

```python
# SALVATAGGIO DEI 3 FILE CLEAN NELLA CARTELLA DATA_CLEAN

serie_a_clean.to_csv(PATH_CLEAN + r"\matches_serie_a_clean.csv", index=False)
premier_clean.to_csv(PATH_CLEAN + r"\matches_premier_league_clean.csv", index=False)
liga_clean.to_csv(PATH_CLEAN + r"\matches_liga_clean.csv", index=False)

print("FILE CLEAN SALVATI NELLA CARTELLA DATA_CLEAN")
```
